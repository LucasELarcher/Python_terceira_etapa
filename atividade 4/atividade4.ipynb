```python
path = r"CD2022_Populacao_2010_Compatibilizada_20231222.xlsx" 
bruto = pd.read_excel(path, header=None) 
```

```python
tabela = bruto.iloc[3:5573, 1:8].copy() 
```

```python
tabela.columns = ["uf","cod_uf","cod_munic","municipio","pop_2010_sinopse","pop_2010_compatibilizada","pop_2022"] 
```

```python
tabela = tabela.reset_index(drop=True)
```

```python
tabela
```

```python
for col in ["cod_uf","pop_2010_sinopse","pop_2010_compatibilizada","pop_2022"]: tabela[col] = pd.to_numeric(tabela[col], errors="coerce")
```

```python
pop_por_estado = ( tabela.groupby("uf")[["pop_2010_compatibilizada", "pop_2022"]] .sum() .reset_index() )
```

```python
pop_por_estado["diferenca"] = pop_por_estado["pop_2022"] - pop_por_estado["pop_2010_compatibilizada"] 
```

```python
pop_por_estado = pop_por_estado.sort_values("diferenca", ascending=False) 
```

```python
pop_por_estado.to_csv("populacao_por_estado.csv", sep=";", index=False)
```

```python
pop_por_estado
```

```python
pop_por_municipio = tabela[
    ["uf", "cod_uf", "cod_munic", "municipio", "pop_2010_compatibilizada", "pop_2022"]
].copy()

pop_por_municipio
```

```python
pop_por_municipio["diferenca"] = pop_por_municipio["pop_2022"] - pop_por_municipio["pop_2010_compatibilizada"]
```

```python
pop_por_municipio = pop_por_municipio.sort_values("diferenca", ascending=False) 
```

```python
pop_por_municipio.to_csv("populacao_por_municipio.csv", sep=";", index=False)
```
