```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path = r"H:\python\3 etapa\atividade 6\Tabela 1.1.1.xls"
path_2012 = r"H:\python\3 etapa\atividade 6\Tabela5-sem_emprego_2012.csv"
path_2026 = r"H:\python\3 etapa\atividade 6\Tabela5-sem_emprego_2026.csv"
```

```python
tabela1 = pd.read_excel(path, header=None)
```

```python
tabela1
```

```python
indicador_1 = tabela1.iloc[8:41].copy()
indicador_1.columns = [
    "Estado",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
indicador_1 = indicador_1.reset_index(drop=True)

for col in indicador_1.columns[1:]:
    indicador_1[col] = pd.to_numeric(indicador_1[col], errors="coerce")

indicador_1
```

```python
regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
estados = indicador_1[~indicador_1["Estado"].isin(regioes)]
estados = estados.reset_index(drop=True)

estados
```

```python
tabela2 = pd.read_csv(path_2012, sep = ";", decimal = ",")
```

```python
tabela2
```

```python
tabela3 = pd.read_csv(path_2026, sep = ";", decimal = ",")
tabela3
```

```python
juntar_tabelas_comp = tabela2.merge(tabela3, on=["Sigla", "Código", "Estado"], how="inner")
```

```python
juntar_tabelas_comp
```

```python

remover = ["total" , "total_branca", "total_preta_parda", "homem_branca", "homem_preta_parda", "mulher_branca", "mulher_preta_parda"]


juntar_tabelas_final = estados.merge(juntar_tabelas_comp, on=["Estado"], how="inner")
juntar_tabelas_final = juntar_tabelas_final[["Sigla", "Código", "Estado"] + [col for col in juntar_tabelas_final.columns if col not in ["Sigla", "Código", "Estado"] + remover]]
```

```python
juntar_tabelas_final
```

```python
ordem = juntar_tabelas_final.sort_values("Desocupados - mulheres (2026 T1)")["Estado"]
longo = juntar_tabelas_final.melt(
    id_vars=["Sigla", "Código", "Estado"],
    value_vars=["Desocupados - mulheres (2012 T1)", "Desocupados - mulheres (2026 T1)"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)
longo["Ano"] = longo["Ano"].map({
    "Desocupados - mulheres (2012 T1)": "2012 T1", 
    "Desocupados - mulheres (2026 T1)": "2026 T1"
})
fig, ax = plt.subplots(figsize=(10, 10))

sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
    palette="muted"
)
ax.axvline(50, color="red", linestyle="--", linewidth=1.5) 
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("Estados")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1", fontsize=14, pad=15)
ax.legend(title="Período Avaliado")
fig.tight_layout()
plt.show()
```
